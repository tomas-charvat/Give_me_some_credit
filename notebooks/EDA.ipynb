```python
import pandas as pd

training_data=pd.read_csv("../data/cs-training.csv")
```

```python


print(training_data.info())
print(training_data.shape)
print(training_data.isna().sum())
print(training_data.isna().mean())

print(training_data["SeriousDlqin2yrs"].sum())
print(training_data["SeriousDlqin2yrs"].mean())

```

```python
training_data.describe().T.reset_index()
```

```python
training_data.groupby(training_data['MonthlyIncome'].isna())['RevolvingUtilizationOfUnsecuredLines'].describe()

```

```python
training_data.groupby(training_data['MonthlyIncome'].isna())['DebtRatio'].describe()
```

```python
training_data['income_missing']=training_data['MonthlyIncome'].isna()

training_data.groupby('income_missing')['SeriousDlqin2yrs'].agg(
    count='count',
    default='sum',
    default_rate='mean'
)
```

```python
training_data['dependents_missing']=training_data['NumberOfDependents'].isna()

training_data.groupby('dependents_missing')['SeriousDlqin2yrs'].agg(
    count='count',
    default='sum',
    default_rate='mean'
)
```
