```python
import pandas as pd
from sklearn.model_selection import train_test_split

training_data= pd.read_csv('../data/cs-training.csv')

X= training_data.drop(columns=['SeriousDlqin2yrs','Unnamed: 0'])

y = training_data['SeriousDlqin2yrs']


X_train, X_test, y_train, y_test = train_test_split(X , y , test_size= 0.2, random_state= 30, stratify=y )




X_train_missing_income = X_train['MonthlyIncome'].isna()
X_test_missing_income = X_test['MonthlyIncome'].isna()

median_income=X_train['MonthlyIncome'].median()

X_train.loc[X_train_missing_income,'DebtRatio'] = (
    X_train['DebtRatio']/median_income
)

X_test.loc[X_test_missing_income,'DebtRatio'] = (
    X_test['DebtRatio']/median_income
)
X_train.loc[X_train_missing_income, 'MonthlyIncome']= median_income
X_test.loc[X_test_missing_income, 'MonthlyIncome'] = median_income

median_dependents= X_train['NumberOfDependents'].median()

X_train['NumberOfDependents']= (
    X_train['NumberOfDependents'].fillna(median_dependents)
)
X_test['NumberOfDependents']= (
    X_test['NumberOfDependents'].fillna(median_dependents)
)

print(X_train.isna().sum())
print(X_test.isna().sum())
```

```python
# 1.test uplne vyhozeni radku s poctem zpozdeni 98 a 96
import numpy as np

# 1. Features + target
X = training_data.drop(
    columns=["SeriousDlqin2yrs", "Unnamed: 0"],
    errors="ignore"
)

y = training_data["SeriousDlqin2yrs"]

# 2. Vyhození podezřelých 96 / 98
late_cols = [
    "NumberOfTime30-59DaysPastDueNotWorse",
    "NumberOfTime60-89DaysPastDueNotWorse",
    "NumberOfTimes90DaysLate"
]

bad_rows = X[late_cols].isin([96, 98]).any(axis=1)

X = X.loc[~bad_rows].copy()
y = y.loc[~bad_rows].copy()

print("Removed rows:", bad_rows.sum())

# 3. Split
X_train, X_valid, y_train, y_valid = train_test_split(
    X,
    y,
    test_size=0.2,
    random_state=30,
    stratify=y
)

# 4. Missing income masks
train_missing_income = X_train["MonthlyIncome"].isna()
valid_missing_income = X_valid["MonthlyIncome"].isna()

# 5. Median income pouze z trainu
median_income = X_train["MonthlyIncome"].median()

# oprava DebtRatio
X_train.loc[train_missing_income, "DebtRatio"] = (
    X_train.loc[train_missing_income, "DebtRatio"] / median_income
)

X_valid.loc[valid_missing_income, "DebtRatio"] = (
    X_valid.loc[valid_missing_income, "DebtRatio"] / median_income
)

# doplnění income
X_train["MonthlyIncome"] = X_train["MonthlyIncome"].fillna(median_income)
X_valid["MonthlyIncome"] = X_valid["MonthlyIncome"].fillna(median_income)

# 6. NumberOfDependents
median_dependents = X_train["NumberOfDependents"].median()

X_train["NumberOfDependents"] = (
    X_train["NumberOfDependents"].fillna(median_dependents)
)

X_valid["NumberOfDependents"] = (
    X_valid["NumberOfDependents"].fillna(median_dependents)
)

# 7. Scaling
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_valid_scaled = scaler.transform(X_valid)

# 8. Logistic regression
log_regression = LogisticRegression(
    l1_ratio=0.0,
    C=1.0,
    max_iter=1000
)

log_regression.fit(X_train_scaled, y_train)

# 9. AUC
y_valid_proba = log_regression.predict_proba(X_valid_scaled)[:, 1]

auc = roc_auc_score(y_valid, y_valid_proba)

print("ROC-AUC:", auc)
# 10. Feature importance / coefficients
coefficients = pd.DataFrame({
    "feature": X_train.columns,
    "coefficient": log_regression.coef_[0],
    "odds_ratio": np.exp(log_regression.coef_[0])
})

coefficients["abs_coefficient"] = coefficients["coefficient"].abs()

coefficients = coefficients.sort_values(
    "abs_coefficient",
    ascending=False
)

print("\nFeature importance:")
print(coefficients)
```

```python

# cross validace pro test stability

from sklearn.model_selection import StratifiedKFold



cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=30
)

auc_scores = []

for train_idx, valid_idx in cv.split(X, y):

    # 1. Aktuální fold
    X_train_cv = X.iloc[train_idx].copy()
    X_valid_cv = X.iloc[valid_idx].copy()

    y_train_cv = y.iloc[train_idx]
    y_valid_cv = y.iloc[valid_idx]


    # 2. Missing income mask
    train_missing_income = X_train_cv["MonthlyIncome"].isna()
    valid_missing_income = X_valid_cv["MonthlyIncome"].isna()


    # 3. Median income POUZE z aktuálního training foldu
    median_income = X_train_cv["MonthlyIncome"].median()


    # 4. Oprava DebtRatio
    X_train_cv.loc[train_missing_income, "DebtRatio"] = (
        X_train_cv.loc[train_missing_income, "DebtRatio"] / median_income
    )

    X_valid_cv.loc[valid_missing_income, "DebtRatio"] = (
        X_valid_cv.loc[valid_missing_income, "DebtRatio"] / median_income
    )


    # 5. Doplnění MonthlyIncome
    X_train_cv["MonthlyIncome"] = (
        X_train_cv["MonthlyIncome"].fillna(median_income)
    )

    X_valid_cv["MonthlyIncome"] = (
        X_valid_cv["MonthlyIncome"].fillna(median_income)
    )


    # 6. Doplnění NumberOfDependents
    median_dependents = X_train_cv["NumberOfDependents"].median()

    X_train_cv["NumberOfDependents"] = (
        X_train_cv["NumberOfDependents"].fillna(median_dependents)
    )

    X_valid_cv["NumberOfDependents"] = (
        X_valid_cv["NumberOfDependents"].fillna(median_dependents)
    )


    # 7. Scaling - fit pouze na training foldu
    scaler = StandardScaler()

    X_train_scaled = scaler.fit_transform(X_train_cv)
    X_valid_scaled = scaler.transform(X_valid_cv)


    # 8. Model
    model = LogisticRegression(
        l1_ratio=0.0,
        C=1.0,
        max_iter=1000
    )

    model.fit(X_train_scaled, y_train_cv)


    # 9. Validation AUC
    probabilities = model.predict_proba(X_valid_scaled)[:, 1]

    auc = roc_auc_score(y_valid_cv, probabilities)
    auc_scores.append(auc)


auc_scores = np.array(auc_scores)

print("Fold AUC:", auc_scores)
print("Mean ROC-AUC:", auc_scores.mean())
print("Std ROC-AUC:", auc_scores.std())
```
